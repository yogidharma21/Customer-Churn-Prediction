## 1. Domain Proyek

### Latar Belakang

*Customer churn* merupakan kondisi ketika pelanggan berhenti menggunakan layanan suatu perusahaan. Dalam industri telekomunikasi yang memiliki banyak pelanggan dan layanan berbasis langganan, mempertahankan pelanggan menjadi salah satu aspek penting dalam menjaga pendapatan dan keberlanjutan bisnis.

Perusahaan perlu memahami karakteristik pelanggan yang memiliki kecenderungan untuk berhenti berlangganan agar dapat mengambil tindakan retensi secara lebih tepat. Dengan memanfaatkan data historis pelanggan, analisis data dan machine learning dapat digunakan untuk mengidentifikasi pola yang berkaitan dengan *customer churn* serta membangun model yang dapat memprediksi kemungkinan pelanggan mengalami churn.

Dataset yang digunakan dalam proyek ini merupakan **Telco Customer Churn Dataset** yang berisi informasi mengenai karakteristik demografis pelanggan, status hubungan dengan pelanggan lain, layanan telekomunikasi yang digunakan, jenis kontrak, metode pembayaran, biaya bulanan, total biaya, serta status churn pelanggan.

Melalui data tersebut, proyek ini bertujuan untuk melakukan analisis terhadap perilaku pelanggan dan mengembangkan model *machine learning* yang dapat memprediksi apakah seorang pelanggan berpotensi mengalami churn. Hasil analisis kemudian digunakan untuk mengidentifikasi faktor-faktor yang paling berkaitan dengan churn dan menghasilkan rekomendasi yang dapat mendukung strategi retensi pelanggan.

### Business Context

**Masalah bisnis:** Perusahaan telekomunikasi dapat mengalami kehilangan pelanggan ketika pelanggan memutuskan untuk berhenti menggunakan layanan. Jika perusahaan tidak mengetahui pelanggan mana yang memiliki risiko churn lebih tinggi, strategi retensi dapat menjadi kurang tepat sasaran dan sumber daya perusahaan dapat digunakan secara tidak efisien.

**Mengapa prediksi churn penting:** Model prediktif dapat membantu perusahaan mengidentifikasi pelanggan yang memiliki kemungkinan lebih tinggi untuk mengalami churn sehingga tim retensi dapat memprioritaskan pelanggan tersebut untuk mendapatkan tindakan yang sesuai.

**Stakeholder yang dapat menggunakan hasil analisis:**
- **Customer Retention/CRM Team** — menentukan pelanggan yang perlu diprioritaskan dalam program retensi.
- **Customer Service Team** — memahami hubungan antara penggunaan layanan dan churn pelanggan.
- **Product Team** — mengevaluasi layanan yang berkaitan dengan tingkat churn.
- **Marketing Team** — merancang strategi penawaran yang lebih terarah.
- **Management** — mendukung pengambilan keputusan berbasis data terkait strategi retensi pelanggan.

**Bagaimana hasil model membantu pengambilan keputusan:** Model dapat menghasilkan prediksi atau probabilitas churn untuk setiap pelanggan. Informasi tersebut dapat digunakan untuk mengelompokkan pelanggan berdasarkan tingkat risiko dan membantu perusahaan menentukan prioritas tindakan retensi secara lebih terukur.

### B. Bivariate Analysis

Categorical Features

```python
for col in categorical_cols:

    churn_rate = (
        pd.crosstab(
            df[col],
            df["Churn"],
            normalize="index"
        ) * 100
    )

    print("=" * 70)
    print(f"Churn Rate by {col}")
    display(churn_rate.round(2))

    plt.figure(figsize=(8, 5))

    churn_yes = (
        df.groupby(col)["Churn"]
        .apply(lambda x: (x == "Yes").mean() * 100)
        .sort_values(ascending=False)
    )

    sns.barplot(
        x=churn_yes.index,
        y=churn_yes.values
    )

    plt.title(f"Churn Rate by {col}")
    plt.xlabel(col)
    plt.ylabel("Churn Rate (%)")
    plt.xticks(rotation=30, ha="right")

    plt.tight_layout()
    plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting set:")
print("X_test :", X_test.shape)
print("y_test :", y_test.shape)
```

Encoding & Scaling

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_cols = ["tenure", "MonthlyCharges"]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numeric_cols
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            ),
            categorical_cols
        )
    ]
)
```

```python
# Fit hanya pada training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data menggunakan preprocessing
# yang sudah dipelajari dari training data
X_test_processed = preprocessor.transform(X_test)

print("Original X_train shape :", X_train.shape)
print("Processed X_train shape:", X_train_processed.shape)

print("\nOriginal X_test shape  :", X_test.shape)
print("Processed X_test shape :", X_test_processed.shape)
```

## 6. Modeling

Target `churn` bersifat **binary classification** (Yes/No). Model yang dibandingkan:
1. **Logistic Regression** — baseline linear.
2. **Decision Tree**
3. **Random Forest**
4. **Gradient Boosting**

```python
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
```

Baseline Model

```python
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    )
}

pipelines = {}

for name, model in models.items():

    pipelines[name] = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])
```

Training

```python
for name, pipeline in pipelines.items():

    print(f"Training {name}...")

    pipeline.fit(
        X_train,
        y_train
    )
```

Evaluasi

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

results = []

for name, pipeline in pipelines.items():

    y_pred = pipeline.predict(X_test)
    y_prob = pipeline.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test,
            y_pred,
            pos_label="Yes"
        ),
        "Recall": recall_score(
            y_test,
            y_pred,
            pos_label="Yes"
        ),
        "F1-Score": f1_score(
            y_test,
            y_pred,
            pos_label="Yes"
        ),
        "ROC-AUC": roc_auc_score(
            (y_test == "Yes").astype(int),
            y_prob
        )
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="ROC-AUC",
    ascending=False
).reset_index(drop=True)

display(
    results_df.style
    .format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}",
        "ROC-AUC": "{:.4f}"
    })
)
```

Threshold Analysis

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds_to_test = [
    0.30,
    0.35,
    0.40,
    0.45,
    0.50,
    0.55,
    0.60
]

threshold_results = []

for threshold in thresholds_to_test:

    y_pred_threshold = np.where(
        y_prob_logreg >= threshold,
        "Yes",
        "No"
    )

    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision_score(
            y_test,
            y_pred_threshold,
            pos_label="Yes"
        ),
        "Recall": recall_score(
            y_test,
            y_pred_threshold,
            pos_label="Yes"
        ),
        "F1-Score": f1_score(
            y_test,
            y_pred_threshold,
            pos_label="Yes"
        )
    })

threshold_df = pd.DataFrame(threshold_results)

display(
    threshold_df.style.format({
        "Threshold": "{:.2f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}"
    })
)
```

```python
plt.figure(figsize=(9, 6))

plt.plot(
    threshold_df["Threshold"],
    threshold_df["Precision"],
    marker="o",
    label="Precision"
)

plt.plot(
    threshold_df["Threshold"],
    threshold_df["Recall"],
    marker="o",
    label="Recall"
)

plt.plot(
    threshold_df["Threshold"],
    threshold_df["F1-Score"],
    marker="o",
    label="F1-Score"
)

plt.xlabel("Classification Threshold")
plt.ylabel("Score")
plt.title("Precision, Recall, and F1-Score by Threshold")
plt.legend()

plt.tight_layout()
plt.show()
```

- Pola trade-off terlihat jelas: Precision naik terus seiring threshold dinaikkan (dari ~0,52 di 0,30 ke ~0,67 di 0,60), sementara Recall justru turun tajam (dari ~0,75 ke ~0,40) — semakin selektif model, semakin banyak pelanggan churn yang terlewat.
- F1-Score relatif stabil di rentang threshold 0,30–0,45 (sekitar 0,61–0,62), lalu mulai menurun setelah threshold 0,45 — menandakan titik keseimbangan terbaik antara Precision dan Recall ada di sekitar rentang ini, bukan di threshold default 0,50.
- Di threshold 0,50 (default), Recall sudah turun ke ~0,58 — artinya kalau tujuannya menjaring sebanyak mungkin pelanggan berisiko churn, threshold default bukan pilihan optimal.
- Pemilihan threshold akhirnya kembali ke prioritas bisnis: kalau biaya menghubungi pelanggan yang salah (false positive) murah, threshold rendah (0,30–0,35) lebih masuk akal; kalau tindakan retensi mahal dan perlu selektif, threshold lebih tinggi (0,50 ke atas) lebih sesuai.

## 8. Model Interpretation

Model yang memiliki performa baik tidak cukup hanya menghasilkan prediksi. Perlu dilakukan interpretasi untuk memahami fitur mana yang paling berpengaruh terhadap prediksi churn.

Pada tahap ini, koefisien **Logistic Regression** digunakan untuk melihat arah dan besar pengaruh masing-masing fitur terhadap kemungkinan pelanggan mengalami churn. Interpretasi ini membantu menghubungkan hasil machine learning dengan kondisi bisnis.

```python
logreg_pipeline = pipelines["Logistic Regression"]

# Ambil preprocessor dari pipeline
fitted_preprocessor = logreg_pipeline.named_steps["preprocessor"]

# Ambil model Logistic Regression
logreg_model = logreg_pipeline.named_steps["model"]

# Ambil nama fitur setelah preprocessing
feature_names = fitted_preprocessor.get_feature_names_out()

# Ambil koefisien
coefficients = logreg_model.coef_[0]

# Buat dataframe
feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients
})

# Nilai absolut untuk melihat besar pengaruh
feature_importance["Abs_Coefficient"] = (
    feature_importance["Coefficient"].abs()
)

# Urutkan berdasarkan nilai absolut
feature_importance = feature_importance.sort_values(
    by="Abs_Coefficient",
    ascending=False
)

display(feature_importance.head(20))
```

```python
top_n = 15

top_features = (
    feature_importance
    .head(top_n)
    .sort_values("Coefficient")
)

plt.figure(figsize=(10, 7))

plt.barh(
    top_features["Feature"],
    top_features["Coefficient"]
)

plt.axvline(
    x=0,
    linestyle="--",
    linewidth=1
)

plt.title("Top Features Influencing Churn Prediction")
plt.xlabel("Logistic Regression Coefficient")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

- **InternetService_Fiber optic** tetap jadi pendorong churn terkuat (koefisien tertinggi, ~1,15), jauh di atas fitur lain — mengonfirmasi pola yang sama seperti sebelumnya.
- **Contract_Two year**, **tenure**, dan **Contract_One year** adalah penahan churn paling kuat (koefisien negatif terbesar) — makin lama tenure dan makin panjang komitmen kontrak, makin kecil kemungkinan churn.
- **MonthlyCharges sekarang konsisten berkoefisien negatif**, sejalan dengan tenure dan contract — setelah TotalCharges dikeluarkan dari model, tidak ada lagi arah koefisien yang saling bertentangan seperti sebelumnya, jadi interpretasinya sekarang lebih bisa dipercaya.
- Fitur value-added seperti **TechSupport** dan **OnlineSecurity** juga tetap menurunkan churn, meski efeknya lebih kecil dibanding tenure/contract.
- Menariknya, fitur layanan tambahan seperti **StreamingTV**, **MultipleLines**, dan **StreamingMovies** justru berkoefisien positif (meningkatkan churn) — kemungkinan pelanggan dengan banyak add-on punya tagihan lebih besar sehingga lebih sensitif terhadap harga, meski ini perlu digali lebih lanjut, bukan disimpulkan sebagai sebab-akibat.

```python
positive_features = (
    feature_importance[
        feature_importance["Coefficient"] > 0
    ]
    .sort_values(
        by="Coefficient",
        ascending=False
    )
    .head(10)
)

negative_features = (
    feature_importance[
        feature_importance["Coefficient"] < 0
    ]
    .sort_values(
        by="Coefficient"
    )
    .head(10)
)

print("Fitur yang meningkatkan kecenderungan prediksi Churn:")
display(
    positive_features[
        ["Feature", "Coefficient"]
    ]
)

print("\nFitur yang menurunkan kecenderungan prediksi Churn:")
display(
    negative_features[
        ["Feature", "Coefficient"]
    ]
)
```
